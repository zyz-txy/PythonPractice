```python
import pandas as pd

#多种创建DataFrame的方式
df1 = pd.DataFrame([
    {"name": "张三", "age": 18}
    ,{"name": "李四", "age": 19}
    ,{"name": "王五", "age": 20}
])

df2 = pd.DataFrame({
    "name": ["张三", "李四", "王五"]
    ,"age": [18, 19, 20]
})

df3 = pd.DataFrame([
    ["张三", 18]
    ,["李四", 19]
    ,["王五", 20]
],columns = ['name','age'])

df4 = pd.DataFrame([
    ["张三", 18]
    ,["李四", 19]
    ,["王五", 20]
],columns = ['name','age'],index=[1,2,3])

#DataFrame中常见属性
#使用tolist转换为列表
# df4.index.tolist()#  索引
# df4.columns.tolist()#  列名
# df4.values.tolist()#  值
# df4.size#  ：元素个数
# df4.shape#  维度：行数和列数
# df4.dtypes#  数据类型

```
