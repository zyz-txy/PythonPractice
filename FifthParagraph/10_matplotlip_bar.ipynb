```python
import matplotlib.pyplot as plt
from matplotlib.axes import Axes
from pydeck.data_utils.viewport_helpers import bbox_to_zoom_level

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']#设置中文为黑体

#创建子图
#figure:画布对象;axes:子图对象
figure, axes = plt.subplots(nrows=1,ncols=2,figsize=(20,6),dpi=100) #nrows代表行,ncols代表列

#图一：柱状图（世界各国石油储备）
countries = ['中国','美国','日本','印度','俄罗斯','英国','德国']
values = [14.5,10.5,9.5,7.5,7.5,4,6]

axes1: Axes = axes[0]
axes1.set_title('世界石油储备',fontsize=18)#设置标签
axes1.bar(countries,values,width=0.5,color='g')#绘制柱状图
axes1.set_xlabel('国家')#y轴标签
axes1.set_ylabel('石油储备（亿吨）')#x轴标签
axes1.grid(linestyle='--',alpha=0.3)#网格线，透明度

#图二：饼状图（世界各国人口） --》擅长描述数据占比
countries2 = ['中国','美国','日本','印度','俄罗斯','英国','德国']
values2 = [1439,331,127,1084,143,67,83]
axes2: Axes = axes[1]               #显示百分比
axes2.pie(values2,labels=countries2,autopct='%.1f%%')
axes2.set_title('世界各国人口',fontsize=18)
#显示图例 ，ncol每行显示的列数，bbox_to_anchor设置图例的位置
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.1))

#保存图片
plt.savefig('data/世界各国石油储备.png',dpi=100)

plt.show()
```
