```python
#数据分组
import pandas as pd
#获取数据
df = pd.read_csv('data/sales.csv',nrows=20)
df['销售金额'] = df['销售数量'] * df['单价']
#分组后求和
df.groupby('产品类别')['订单号'].count()#缺失值不参与统计，可以先填充
#分组后统计数量
df.groupby('产品类别')['销售数量'].sum()
#根据产品类别分组，统计各个类别的销售金额之和
df.groupby('产品类别')['销售金额'].sum()
#分组后求最大值
df.groupby('产品类别')['销售金额'].max()
#分组后求最小值
df.groupby('产品类别')['销售金额'].min()
#分组后求平均值
df.groupby('产品类别')['销售金额'].mean()
#agg（聚合操作）,一次性求多个值
df.groupby('产品类别').agg({'销售金额':['max','min','mean']})
df.groupby('产品类别')['销售金额'].agg(['sum','max','min'])
df.groupby('产品类别').agg({'销售数量' :'sum','销售金额' :'sum','单价':'mean'})
```
