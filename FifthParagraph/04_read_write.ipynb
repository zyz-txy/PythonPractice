```python
from logging import info

#基于Pandas中提供的API，可以很方便的实现各类数据文件（csv,excel,数据库,网络数据等）的读取和写入

import pandas as pd

#读取数据
df = pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])

#数据处理
df['销售额'] = df['销售数量'] * df['单价']

#写入数据--->to_scv   index=False(不写入索引列)
df.to_csv('data/sales_01.csv',index=False)
```

```python
#数据查看 --head,tail,describe,info,shape,columns
df1 = pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])
df1.head(10)#显示前N行数据
df1.tail(10)#显示后N行数据
df1.describe()#显示统计信息
df1.info()#显示数据信息（列名，非空计数，数据类型）
df1.shape#显示数据形状（行数，列数）
df1.columns#显示列名
```

```python
#数据选择
df2 = pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])
#选择列
df2['订单号'] #单列
df2[['订单号','产品名称','单价']] #多列
#选择行-->iloc,loc
# iloc---->基于行号选择行，语法：df.iloc[start:stop:step] , 取头不取尾
df2.iloc[0:5:1]#行号和索引不是同一个东西
#行号不显示，索引为第一列
df2.loc[1:5:2]#基于行索引选择行
```
