```python
from logging import info

#基于Pandas中提供的API，可以很方便的实现各类数据文件（csv,excel,数据库,网络数据等）的读取和写入

import pandas as pd

#读取数据
df = pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])

#数据处理
df['销售额'] = df['销售数量'] * df['单价']

#写入数据--->to_scv   index=False(不写入索引列)
df.to_csv('data/sales_01.csv',index=False)
```

```python
import pandas as pd
df3 = pd.read_csv('data/sales.csv')

#配置项，展示所有的数据
pd.set_option('display.max_rows',None)

#数据过滤
#1.获取1销售数量>=10的订单数据
df3[df3['销售数量'] >= 10]
#获取单价在100-200之间的订单数据
df3[df3['单价'].between(100,200,inclusive="neither")]#默认包含边界值
#inclusive 参数 right left neither
df3[(df3['单价']>100) & (df3['单价']<200)]
#获取产品类别为食品或图书的订单数据
df3[df3['产品类别'].isin(['食品','图书'])]
#获取销售数量大于8且单价大于等于100的订单数据  并且&   或者|
df3[(df3['销售数量']>8) & (df3['单价']>=100)]
#获取产品类别为服装或食品，支付方试为支付宝或完微信支付的订单数据
df3[(df3['产品类别'].isin(['服装','食品'])) & (df3['支付方式'].isin(['支付宝','微信支付']))]
```
