```python
import matplotlib.pyplot as plt
import random

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

x = [i for i in range(1,25)]
y = [ random.randint(10,15) for i in x]

#在plot之前创建画布
plt.figure(figsize = (10,5)) #设置画布大小，宽，高
plt.plot(x,y) #折线图 -->如果没有画布，会自动创建画布

#设置折线图详细信息
plt.title("气温变化折线图",fontsize = 15) #标题
plt.xlabel("时间") #x轴
plt.ylabel("温度") #y轴
plt.xticks(x[1::2])

plt.show() #显示图表
```
