```python
#重复值处理，缺失值处理，异常值处理，异常格式处理

import pandas as pd

pd.set_option('display.max_columns', 30)

#1读取数据
df = pd.read_csv('data/sales.csv')

#缺失值处理
df.isnull() # 查看缺失值
#删除缺失值所在行
df.dropna()
#删除缺失值所在列
df = df.dropna(axis = 1)#调用dropna方法会返回一个新的DataFrame
#填充缺失值
df.fillna('--')
df.ffill() #使用上一行数据填充
df.bfill() #使用下一行数据填充
```

```python
#重复值处理
#查看重复值
df.duplicated()  #所有列的数据全相同才重复
df.duplicated(subset = ['订单号']) #指定列的数据相同才重复，可以多列
#删除重复值, first保留第一行 ,last保留最后一行,false均不保留
df.drop_duplicates(subset = ['订单号'],keep = 'first')
```

```python
#异常值处理
#查看异常值
df[df['单价'] < 0]
#删除异常值
df.drop(df[df['单价'] < 0].index)
#修复异常值
df['单价'] = df['单价'].abs() #绝对值
df
```

```python
#格式异常处理
df['订单日期'] = df['订单日期'].str.replace('/','-')#将订单日期里面的/转换为-
df
```
